```python
%reset -f
import os
import matplotlib.pyplot as plt

# 1. Création automatique du dossier data/raw et du fichier data.txt
os.makedirs("data/raw", exist_ok=True)
with open("data/raw/data.txt", "w", encoding="utf-8") as f:
    f.write("Laptop, 1000,3\nPhone, 500,5\nTablet, 700,2\nMonitor,300,4\nCamera,400,1")

# 2. Fonction pour nettoyer les lignes (Question 01)
def clean_lines(lines):
    return [line.strip() for line in lines if line.strip()]

# 3. Fonction pour analyser les produits (Question 02)
def parse_products(lines):
    parsed = []
    for line in lines:
        parts = line.split(',')
        if len(parts) == 3:
            name = parts[0].strip()
            price = float(parts[1].strip())
            quantity = int(parts[2].strip())
            parsed.append((name, price, quantity))
    return parsed

# 4. Lecture du fichier et calcul des ventes (Question 04)
with open("data/raw/data.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

cleaned_lines = clean_lines(lines)
parsed_data = parse_products(cleaned_lines)

total_sales = 0.0
product_names = []
product_sales = []

for name, price, quantity in parsed_data:
    val = price * quantity
    total_sales += val
    product_names.append(name)
    product_sales.append(val)
    print(f"{name}: {val}")

print(f"Total Sales: {total_sales}")

# 5. Visualisation graphique / Bar Chart (Question 05)
plt.figure(figsize=(8, 5))
plt.bar(product_names, product_sales, color='skyblue')
plt.xlabel('Noms des produits')
plt.ylabel('Ventes totales')
plt.title('Ventes totales par produit')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
